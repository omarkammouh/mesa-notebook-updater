# Sheep on a Pasture

A minimal grid model: a flock of sheep wander at random across a toroidal
pasture. Each step every sheep looks at its neighbouring cells and moves to one
of them. Nothing is eaten yet — this is only the movement skeleton.

```python
import mesa
from mesa.space import MultiGrid
```

```python
class Sheep(mesa.Agent):
    def __init__(self, model):
        super().__init__(model)

    def step(self):
        neighborhood = self.model.grid.get_neighborhood(
            self.pos, moore=True, include_center=False
        )
        new_pos = self.model.random.choice(neighborhood)
        self.model.grid.move_agent(self, new_pos)
```

```python
class Pasture(mesa.Model):
    def __init__(self, n=20, width=10, height=10, seed=None):
        super().__init__(seed=seed)
        self.grid = MultiGrid(width, height, torus=True)
        for i in range(n):
            a = Sheep(self)
            x = self.random.randrange(width)
            y = self.random.randrange(height)
            self.grid.place_agent(a, (x, y))

    def step(self):
        for a in list(self.agents):
            a.step()
```

```python
model = Pasture(seed=1)
for _ in range(10):
    model.step()

print("steps run:", model.steps, "| sheep:", len(model.agents))
```

The flock has wandered for ten steps. From here you could add grass to the
cells and let the sheep graze, then reproduce — but the movement loop above is
the core of every grid model.
