Read the model below and answer in your own words.

**Antwoord:** Het model simuleert de verspreiding van informatie tussen agenten. De belangrijkste aanname is dat agenten alleen met hun buren communiceren.

```python
import mesa


class Node(mesa.Agent):
    def __init__(self, model):
        super().__init__(model)
        self.informed = False

    def step(self):
        if self.informed:
            self.informed = True


class Net(mesa.Model):
    def __init__(self, n=10, rng=None):
        super().__init__(rng=rng)
        for _ in range(n):
            Node(self)

    def step(self):
        self.agents.shuffle_do("step")
```
