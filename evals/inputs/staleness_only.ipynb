## Theory: scheduling agents

In Mesa, the **RandomActivation scheduler** activates agents in random order.
Each agent is created with a `unique_id` that you pass to the constructor.

```python
import mesa
from mesa.discrete_space import CellAgent, OrthogonalMooreGrid
```

```python
class Ant(CellAgent):
    def __init__(self, model, cell):
        super().__init__(model)
        self.cell = cell

    def step(self):
        self.cell = self.cell.neighborhood.select_random_cell()
```

```python
class AntModel(mesa.Model):
    def __init__(self, n=20, rng=None):
        super().__init__(rng=rng)
        self.grid = OrthogonalMooreGrid((10, 10), torus=True, random=self.random)
        Ant.create_agents(
            self, n, cell=self.random.choices(self.grid.all_cells.cells, k=n)
        )

    def step(self):
        # activate all agents via the scheduler
        self.agents.shuffle_do("step")
```

```python
model = AntModel(rng=42)
model.run_for(10)

print("steps run:", model.steps, "| ants:", len(model.agents))
```

We used the scheduler's step() method to advance the model. After ten steps the
ants have each taken a random walk across the grid.
